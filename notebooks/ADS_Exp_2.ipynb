```python
print("Before:", df.shape)

df = df.drop_duplicates().copy()

print("After:", df.shape)
```

```python
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

categorical_cols = df.select_dtypes(include="object").columns

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

print(df.isnull().sum())
```

```python
df["date"] = pd.to_datetime(df["date"], errors="coerce")
```

```python
df["Year"] = df["date"].dt.year
df["Month"] = df["date"].dt.month
df["Quarter"] = df["date"].dt.quarter
df["Day"] = df["date"].dt.day
df["DayOfWeek"] = df["date"].dt.day_name()
df["IsWeekend"] = df["date"].dt.weekday >= 5

# Since there is no time column
df["TimeOfDay"] = "Unknown"

df.head()
```

```python
df.to_csv("cleaned_layoffs.csv", index=False)

print("Dataset Saved Successfully!")
```

```python
import pandera as pa
from pandera import Column, DataFrameSchema, Check

schema = DataFrameSchema({
    "company": Column(str, nullable=False),
    "country": Column(str, nullable=False),
    "total_laid_off": Column(float, Check.ge(0), nullable=True),
    "year": Column(int, Check.in_range(2000, 2035), nullable=True)
})

validated_df = schema.validate(df)

print("✅ Dataset validation successful!")
```

```python
validated_df.to_csv("validated_cleaned_layoffs.csv", index=False)
```
