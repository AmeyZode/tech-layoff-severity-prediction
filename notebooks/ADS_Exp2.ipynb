```python
df = df.drop(columns=[c for c in df.columns if c.startswith("unnamed")], errors="ignore")

df = df.rename(columns={
    "funds_raised_m": "funds_raised_mil",
    "_laid_off": "percentage_laid_off",
})

print(df.columns.tolist())
```

```python
text_cols = ["company", "city", "industry", "stage", "status"]
for col in text_cols:
    df[col] = df[col].astype("string").str.strip()
```

```python
country_map = {
    "usa": "United States",
    "u.s.": "United States",
    "u.s.a.": "United States",
    "united states of america": "United States",
    "us": "United States",
    "n/a": "Unknown",
    "-": "Unknown",
    "": "Unknown",
}

df["country"] = df["country"].astype("string").str.strip().str.lower().map(
    lambda v: country_map.get(v, v)
)


df["country"] = df["country"].map(lambda v: v if pd.isna(v) else (str(v).title() if v != "unknown" else v))

df.loc[df["country"].isin(["United States", "united states"]), "country"] = "United States"

df["country"].value_counts().head(10)
```

```python
def clean_percentage(v):
    if pd.isna(v):
        return np.nan
    if isinstance(v, str):
        v = v.strip().rstrip("%")
        try:
            v = float(v)
        except ValueError:
            return np.nan
    return float(v)

df["percentage_laid_off"] = df["percentage_laid_off"].map(clean_percentage)

for col in ["company_size_before", "company_size_after", "total_laid_off", "funds_raised_mil"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

df[["percentage_laid_off", "company_size_before", "company_size_after",
    "total_laid_off", "funds_raised_mil"]].describe()
```

```python
df["date"] = pd.to_datetime(df["date"], format="mixed", errors="coerce")
df["date_added"] = pd.to_datetime(df["date_added"], format="mixed", errors="coerce")

df[["date", "date_added"]].head()
```

```python
before = len(df)
df = df.drop_duplicates(subset=[c for c in df.columns if c != "record_id"])
after = len(df)
print(f"dropped {before - after} duplicate rows ({before} -> {after})")
```

```python
df.loc[df["total_laid_off"] < 0, "total_laid_off"] = np.nan
df.loc[df["percentage_laid_off"] > 100, "percentage_laid_off"] = np.nan
df.loc[df["funds_raised_mil"] > 200_000, "funds_raised_mil"] = np.nan
df.loc[df["company_size_after"] > df["company_size_before"] * 10, "company_size_after"] = np.nan

df[["total_laid_off", "percentage_laid_off", "funds_raised_mil"]].describe()
```

```python
df["city"] = df["city"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
df["industry"] = df["industry"].fillna("Unknown")
df["stage"] = df["stage"].fillna("Unknown")
df["status"] = df["status"].fillna("Unknown")

df["company_size_after"] = df["company_size_after"].fillna(df["company_size_before"])
df["total_laid_off"] = df["total_laid_off"].fillna(df["total_laid_off"].median())
df["percentage_laid_off"] = df["percentage_laid_off"].fillna(df["percentage_laid_off"].median())
df["funds_raised_mil"] = df["funds_raised_mil"].fillna(df["funds_raised_mil"].median())

df = df.dropna(subset=["date"])

missing_id = df["record_id"].isna()
df.loc[missing_id, "record_id"] = [f"layoff_new_{i:04d}" for i in range(missing_id.sum())]

print("missing values left:")
print(df.isna().sum())
```

```python
!pip install great_expectations -q
import great_expectations as gx

EXPECTED_COLUMNS = [
    "record_id", "company", "city", "country", "is_us", "industry", "stage",
    "company_size_before", "company_size_after", "total_laid_off",
    "percentage_laid_off", "funds_raised_mil", "date", "year", "date_added", "status",
]

context = gx.get_context(mode="ephemeral")
data_source = context.data_sources.add_pandas("layoffs_pandas")
data_asset = data_source.add_dataframe_asset(name="layoffs_cleaned_final")
batch_definition = data_asset.add_batch_definition_whole_dataframe("full_dataset")

suite = gx.ExpectationSuite(name="layoffs_cleaned_final_suite")

expectations = [
    gx.expectations.ExpectTableColumnsToMatchSet(column_set=EXPECTED_COLUMNS, exact_match=True),
    gx.expectations.ExpectColumnValuesToNotBeNull(column="record_id"),
    gx.expectations.ExpectColumnValuesToBeUnique(column="record_id"),
    gx.expectations.ExpectColumnValuesToNotBeNull(column="company"),
    gx.expectations.ExpectColumnValuesToNotBeNull(column="country"),
    gx.expectations.ExpectColumnValuesToBeOfType(column="is_us", type_="bool"),
    gx.expectations.ExpectColumnValuesToBeBetween(column="percentage_laid_off", min_value=0, max_value=100),
    gx.expectations.ExpectColumnValuesToBeBetween(column="total_laid_off", min_value=0, max_value=None),
    gx.expectations.ExpectColumnValuesToBeBetween(column="company_size_before", min_value=0, max_value=None),
    gx.expectations.ExpectColumnValuesToBeBetween(column="company_size_after", min_value=0, max_value=None),
    gx.expectations.ExpectColumnValuesToBeBetween(column="funds_raised_mil", min_value=0, max_value=None),
    gx.expectations.ExpectColumnValuesToBeInSet(column="status", value_set=["Public", "Private", "Unknown"]),
    gx.expectations.ExpectColumnValuesToBeBetween(column="year", min_value=2018, max_value=2030),
]
for exp in expectations:
    suite.add_expectation(exp)

context.suites.add(suite)

validation_definition = gx.ValidationDefinition(
    name="layoffs_cleaned_final_validation", data=batch_definition, suite=suite
)
result = validation_definition.run(batch_parameters={"dataframe": df})

print("all expectations passed:", result.success)
```
